A ball on the end of a string is whirled around in a horizontal
circle of radius 0.300 m. The plane of the circle
is 1.20 m above the ground. The string breaks and the
ball lands 2.00 m (horizontally) away from the point on
the ground directly beneath the ball’s location when the
string breaks. Find the radial acceleration of the ball
during its circular motion.


```python
r = 0.300 * u.m
h = 1.20 * u.m
g = 9.8 * (u.m / u.s**2)
d = 2.00 * u.m  # distance ball travels horizontally

# Time of Flight (t)
# When the string breaks, it becomes a projectile with an initial horizontal
# velocity (v) and zero initial vertical velocity. The time it takes to hit the
# ground depends only on the height = 1.20 and gravity g
# h = 1/2 * g * t^2 ==> t = (2 h / g)^(1/2)
t = (2 * h / g) ** (1 / 2)

# Horizontal Velocity (v)
# The ball travels a horizontal distance (d) during the time (t).
v = d / t

# Radial Acceleration (a_r)
a_r = v**2 / r
print(f"Radial acceleration a_r={a_r:.1f}")
```

#### Problem 39
page 104

Why is the following situation impossible? Ronald Acuna
Jr. hits a home run so that the baseball just clears the top
row of bleachers, 24.0 m high, located 130 m from home
plate. The ball is hit at 41.7 m/s at an angle of 35.0° to the
horizontal, and air resistance is negligible.

```python
h = 24.0 * u.m
D = 130 * u.m
speed = 41.7 * (u.m / u.s)
theta = vp.radians(35.0)
g = -9.8 * (u.m / u.s**2)
y0 = 1.00 * u.m


# Horizontal velocity
v_x = speed * vp.cos(theta)
# Vertical velocity
v_y = speed * vp.sin(theta)

# Time to the clear the Distance
# v = D / t ==> t = D / V
t = D / v_x
print(f"Time to reach the top row of bleachers: {t:.1f}")

# Calculate the height at that time
# y = y0 + v_y * t + 1/2 * a * t^2
y = y0 + (v_y * t) + ((1 / 2) * g * t**2)
print(f"The height of the ball at that distance is {y:.2f}")
print(f"So, the ball is {h - y:.2f} lower than the bleachers")
```

#### Problem 41

An astronaut on the surface of the Moon fires a cannon
to launch an experiment package, which leaves the
barrel moving horizontally. Assume the free-fall acceleration
on the Moon is one-sixth of that on the Earth.


**(a)** 

What must the muzzle speed of the package be so
that it travels completely around the Moon and returns
to its original location? 

```python
g_earth = 9.80 * (u.m / u.s**2)
g_moon = 1 / 6 * g_earth
r_moon = 1.74e6 * u.m

# Moon's gravitational acceleration is the probe's centripetal acceleration
# a = v^2 / r ==> v = (a * r)**(1/2)
v = (g_moon * r_moon) ** (1 / 2)
print(f"The speed v={v.to(u.km / u.s):.2f}")
```

(b) What time interval does this
trip around the Moon require?

```python
# Time interval
# V = 2*pi*r / T ==> T = (2*pi*r) / v
T = (2 * vp.pi * r_moon) / v
print(f"The time interval T = {T.to(u.h):.2f}")
```

#### Problem 48
page 105

You are on the Pirates of the Caribbean attraction in
the Magic Kingdom at Disney World. Your boat rides
through a pirate battle, in which cannons on a ship and
in a fort are firing at each other. While you are aware
that the splashes in the water do not represent actual
cannonballs, you begin to wonder about such battles
in the days of the pirates. Suppose the fort and the ship
are separated by 75.0 m. You observe that the cannons
in the fort are aimed so that their cannonballs would
be fired horizontally from a height of 7.00 m above
the water. 

**(a)** 

You wonder at what speed they must be
fired in order to hit the ship before falling in the water.

```python
x = 75.0 * u.m
y0 = 7 * u.m
g = -9.8 * (u.m / u.s**2)

# Calculate the time as the ball drops straight (-y) to the water
# When the ball reach the water y = 0
# y = y0 + 1/2 * g * t^2 ==> t = (-2*y0 / g)**(1/2)
t = (-2 * y0 / g) ** (1 / 2)
print(f"The time t={t:.3f}")

# In the same time the ball should clear the distance to the ship
v = x / t
print(f"The fired speed v={v:.1f}")
```


**(b)**  

Then, you think about the sludge that must build up
inside the barrel of a cannon. This sludge should slow
down the cannonballs. A question occurs in your mind:
if the cannonballs can be fired at only 50.0% of the
speed found earlier, is it possible to fire them upward at
some angle to the horizontal so that they would reach
the ship?

```python
# 50% of v
v_i = 1 / 2 * v.value
x_target = 75.0
h = 7.0
g = 9.8

theta = sp.symbols("theta")
# Range equation for a projectile launched from height h
range_eq = (v_i * sp.cos(theta) / g) * (
    v_i * sp.sin(theta) + sp.sqrt((v_i * sp.sin(theta)) ** 2 + 2 * g * h)
) - x_target

# Solve for theta(searching near 20-30 degrees)
angle_rad = sp.nsolve(range_eq, theta, 0.4)
print(f"Required angle: {round(float(sp.deg(angle_rad)), 2)} degrees")
```
