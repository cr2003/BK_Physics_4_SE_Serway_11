#### Problem 7

page 101

Mayan kings and many school sports teams are named for the puma, cougar, or 
mountain lion—Felis concolor—the best jumper among animals. It can jump to a 
height of 12.0 ft when leaving the ground at an angle of 45.0°. With what speed 
in SI units, does it leave the ground to make this leap?


```python
# 1. SYMBOLIC STAGE (SymPy)
y, v_f, v_i, theta, a = sp.symbols("y v_f v_i theta a", real=True)

# Define the kinematic equation: v_f^2 = (v_i * sin(theta))^2 + 2 * a * y
equation = sp.Eq(v_f**2, (v_i * sp.sin(theta)) ** 2 + 2 * a * y)

# Solve for initial velocity v_i
solutions = sp.solve(equation, v_i)
v_i_isolated = solutions[1]  # Select positive root

print("--- high Formula Solved by SymPy ---")
display(Math(rf"{sp.latex(sp.Eq(v_i, v_i_isolated))}"))
```

```python
# 2. NUMERICAL EVALUATION WITH ASTROPY UNITS

# Given parameters
h_val = 12.0 * u.imperial.foot
h_val = h_val.to(u.m)

# Use np.radians or pure float for numerical evaluation in astropy
theta_val = np.radians(45.0)
v_f_val = 0.0 * (u.m / u.s)
g = -9.80 * (u.m / u.s**2)
a_val = g

# Pass symbolic variable 'theta' to lambdify (NOT theta_val)
v_i_func = sp.lambdify((a, y, v_f, theta), v_i_isolated)

# Pass the numerical value 'theta_val' when calling the function
v_i_val = v_i_func(a_val, h_val, v_f_val, theta_val)

print("--- AstroPy Numerical Result ---")
print(f"Speed v_i: {v_i_val:.2f} equivalent to {v_i_val.to(u.km / u.h):.0f}")
```

#### Problem 15

page 102

A home run is hit in such a way that the baseball just
clears a wall 21.0 m high, located 130 m from home
plate. The ball is hit at an angle of 35.0° to the horizontal,
and air resistance is negligible. 

(Assume the ball is
hit at a height of 1.00 m above the ground.)

Find 

**(a)** 

the initial speed of the ball, 



**(b)** 

the time it takes the ball to reach the wall

**(c)** 

the velocity components and the speed
of the ball when it reaches the wall. 

```python
# =====================================================================
# 1. SYMBOLIC STAGE (SymPy)
# =====================================================================

# Define abstract mathematical symbols
v0, theta, g, x, y, y0, t = sp.symbols("v0 theta g x y y0 t", real=True, positive=True)

# Equation 1: Time of flight to reach horizontal distance x
# x = v0 * cos(theta) * t  ==>  t = x / (v0 * cos(theta))
t_expr = x / (v0 * sp.cos(theta))

# Equation 2: Vertical position at time t
y_eq = sp.Eq(y, y0 + v0 * sp.sin(theta) * t_expr - sp.Rational(1, 2) * g * t_expr**2)

# Solve for initial speed v0 squared, then take square root
v0_sq_solutions = sp.solve(y_eq, v0**2)
v0_sq_expr = v0_sq_solutions[0]
v0_expr = sp.sqrt(v0_sq_expr)

print("--- Initial Speed Formula Solved by SymPy ---")
display(Math(rf"v_0 = {sp.latex(v0_expr)}"))

print("--- Time of Flight Formula ---")
display(Math(rf"t = {sp.latex(t_expr)}"))
```

```python
# =====================================================================
# 2. NUMERICAL EVALUATION WITH ASTROPY UNITS
# =====================================================================

# Given parameters with AstroPy units
x_val = 130.0 * u.m
y_val = 21.0 * u.m
y0_val = 1.00 * u.m
theta_deg = 35.0 * u.deg
theta_rad = theta_deg.to(u.rad).value  # Numerical float for lambdify
g_val = 9.80 * (u.m / u.s**2)

# Convert SymPy expressions to callable functions
v0_func = sp.lambdify((x, y, y0, theta, g), v0_expr, "numpy")
t_func = sp.lambdify((x, v0, theta), t_expr, "numpy")

# (a) Calculate initial speed v0
v0_val = v0_func(x_val, y_val, y0_val, theta_rad, g_val)

# (b) Calculate time to reach wall t
t_val = t_func(x_val, v0_val, theta_rad)

# (c) Calculate velocity components and total speed at the wall
vx_val = v0_val * np.cos(theta_rad)
vy_val = v0_val * np.sin(theta_rad) - g_val * t_val
speed_val = np.sqrt(vx_val**2 + vy_val**2)

print("\n--- AstroPy Numerical Results ---")
print(f"(a) Initial speed (v0): {v0_val:.3f}")
print(f"(b) Time to reach wall (t): {t_val:.3f}")
print(f"(c) Horizontal velocity at wall (vx): {vx_val:.3f}")
print(f"    Vertical velocity at wall (vy): {vy_val:.3f}")
print(f"    Speed at wall: {speed_val:.3f}")
```

#### Problem 17

page 102

A boy stands on a diving board and tosses a stone into
a swimming pool. The stone is thrown from a height of
2.50 m above the water surface with a velocity of 4.00 m/s
at an angle of 60.0° above the horizontal. As the stone
strikes the water surface, it immediately slows down to
exactly half the speed it had when it struck the water and
maintains that speed while in the water. After the stone
enters the water, it moves in a straight line in the direction
of the velocity it had when it struck the water. If the pool
is 3.00 m deep, how much time elapses between when the
stone is thrown and when it strikes the bottom of the pool?


```python
# =====================================================================
# 1. SYMBOLIC STAGE (SymPy)
# =====================================================================

# Define symbolic variables
v0, theta, g, h, d = sp.symbols("v0 theta g h d", real=True, positive=True)

# 1. Initial vertical velocity component
v0y = v0 * sp.sin(theta)

# 2. Vertical velocity component right before hitting the water (downward, so negative)
v1y = -sp.sqrt(v0y**2 + 2 * g * h)

# 3. Total speed when striking the water
v1 = sp.sqrt(v0**2 + 2 * g * h)

# 4. Time in air (t1)
t1 = (v0y - v1y) / g

# 5. Time underwater (t2)
# Since vertical velocity underwater is half of |v1y|, t2 = d / (0.5 * |v1y|)
t2 = 2 * d / sp.Abs(v1y)

# 6. Total elapsed time
t_total = t1 + t2

print("--- Symbolic Formula for Time in Air (t1) ---")
display(Math(rf"t_1 = {sp.latex(t1)}"))

print("--- Symbolic Formula for Time Underwater (t2) ---")
display(Math(rf"t_2 = {sp.latex(t2)}"))

print("--- Total Time Symbolic Expression ---")
display(Math(rf"t_{{total}} = {sp.latex(sp.simplify(t_total))}"))
```

```python
# -----------------------------------------------------------------
# 2. Numerical Evaluation with AstroPy Units
# -----------------------------------------------------------------


# Defined input values with physical units
h_val = 2.50 * u.m
v0_val = 4.00 * (u.m / u.s)
theta_deg = 60.0 * u.deg
theta_rad = theta_deg.to(u.rad).value
d_val = 3.00 * u.m
g_val = 9.80 * (u.m / u.s**2)

# Lambdify the symbolic expressions from Phase 1
v0, theta, g, h, d = sp.symbols("v0 theta g h d", real=True, positive=True)

v0y_expr = v0 * sp.sin(theta)
v1y_expr = -sp.sqrt(v0y_expr**2 + 2 * g * h)
t1_expr = (v0y_expr - v1y_expr) / g
t2_expr = 2 * d / sp.Abs(v1y_expr)

t1_func = sp.lambdify((v0, theta, g, h), t1_expr, "numpy")
t2_func = sp.lambdify((v0, theta, g, h, d), t2_expr, "numpy")

# Numerical calculation
t1_val = t1_func(v0_val, theta_rad, g_val, h_val)
t2_val = t2_func(v0_val, theta_rad, g_val, h_val, d_val)
t_total_val = t1_val + t2_val

print("--- AstroPy Numerical Results ---")
print(f"Time in air (t1)       : {t1_val:.4f}")
print(f"Time underwater (t2)   : {t2_val:.4f}")
print(f"Total elapsed time (t) : {t_total_val:.4f}")
```
