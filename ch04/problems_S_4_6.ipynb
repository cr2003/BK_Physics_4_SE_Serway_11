```python
import vpython as vp
from astropy import units as u
from astropy import constants as const
import sympy as sp
import numpy as np
from IPython.display import display, Math
```

```python
def fmt_vec(v, precision=2, type="f"):
    """Format the output of a vector with the specified precision"""
    return f"<{v.x:.{precision}{type}}, {v.y:.{precision}{type}}, {v.z:.{precision}{type}}>"
```

```python
def math_to_bearing(theta):
    """Converts mathematical angle (degrees) to navigation bearing (degrees)."""
    return (90 - theta) % 360
```

#### Problem 26

page 103

The pilot of an airplane notes that the compass indicates
a heading due west. The airplane’s speed relative
to the air is 150 km/h. The air is moving in a wind at
30.0 km/h toward the north. Find the velocity of the
airplane relative to the ground.

```python
# Define vectors (East is +x, North is +y)
v_pa = vp.vec(-150, 0, 0)  # Plane relative to Air
v_ag = vp.vec(0, 30, 0)  # Air relative to Ground (Wind)

# Ground velocity
v_pg = v_pa + v_ag

# Calculations
ground_speed = vp.mag(v_pg)

# Calculate math angle in degrees
math_angle = vp.degrees(vp.atan2(v_pg.y, v_pg.x))

# Convert to navigation bearing
bearing = math_to_bearing(math_angle)

print(f"Ground Velocity: {v_pg} k/m")
print(f"Ground Speed: {ground_speed:.0f} k/m")
print(f"Navigation Bearing: {bearing:.1f} degrees")
```

You are taking flying lessons from an experienced pilot.

pilot seat. The control tower radios the plane, saying
that, while you have been airborne, a 25-mi/h crosswind
has arisen, with the direction of the wind perpendicular
to the runway on which you plan to land. The
pilot tells you that your normal airspeed as you land
will be 80 mi/h. This speed is relative to the air, in the
direction in which the nose of the airplane points. He
asks you to determine the angle at which the aircraft
must be “crabbed,” that is, the angle between the centerline
of the aircraft and the centerline of the runway that
will allow the airplane’s velocity relative to the ground
to be parallel to the runway.

<img src="..//pics/Fig_ch4_problem_27.png" width=600> 
<img src="..//pics/Fig_ch4_problem_27_correction.png" width=600> 





NB: The vector of the wind (V_AG) is inverted, should be like the second diagram.

```python
# Define abstract mathematical symbols
v_pa, v_ag, theta = sp.symbols("v_pa, v_ag, theta", real=True, positive=True)

# Vector equation component perpendicular to runway:
# v_pa * sin(theta) = v_ag
eq = sp.Eq(v_pa * sp.sin(theta), v_ag)

# Solve for sin(theta)
sin_theta_expr = v_ag / v_pa

# Solve for crab angle theta
theta_expr = sp.asin(sin_theta_expr)

print("--- Crab Angle Formula Solved by SymPy ---")
display(Math(rf"\theta = {sp.latex(theta_expr)}"))
```

```python
# =====================================================================
# NUMERICAL COMPUTATION WITH ASTROPY UNITS
# =====================================================================

# Given values
v_wind = 25.0 * (u.imperial.mi / u.h)  # wind
v_air = 80.0 * (u.imperial.mi / u.h)  # plane

# Calculate Crab Angle
crab_angle_rad = vp.asin(v_wind / v_air)
crab_angle_deg = vp.degrees(crab_angle_rad)

# Calculate Ground Speed (the adjacent side of the triangle)
v_ground = v_air * vp.cos(crab_angle_rad)

print(f"Crab Angle: {crab_angle_deg:.2f} degrees into the wind")
print(f"Resulting Ground Speed: {v_ground:.2f}")
```

#### Problem 28

A car travels due east with a speed of 50.0 km/h. Raindrops
are falling at a constant speed vertically with
respect to the Earth. The traces of the rain on the side
windows of the car make an angle of 60.0° with the vertical.
Find the velocity of the rain with respect to (a) the
car and (b) the Earth.

<img src="..//pics/Fig_ch4_problem_28.png" width=300> 


```python
# Variables
# v_ce (velocity car relative to the eart)
# v_wc (velocity water relative to the car)
# v_we (velocity water relative to the earth)
# theta (angle with vertical)

# given values
v_ce = vp.vec(50.0, 0, 0)
theta = vp.radians(60.0)  # angle with vertical

# (a)
# From geometry:
# Since v_we is vertical, w_wc sin(theta) = v_ce ==> w_wc = v_ce / sin(theta)
v_wc = v_ce / vp.sin(theta)


# (b)
# Since v_ce has zero vertical component, w_we = v_wc cos(theta)

w_we = v_wc * vp.cos(theta)


print(f"(a) Rain speed relative to car w_wc_mag: {w_wc.mag:.1f} km/h")
print(f"with angle theta {vp.degrees(theta):.0f} degrees west of vertical")
print(f"(b) Rain speed relative to Earth: {w_we.mag:.1f} km/h downward")
```

#### Problem 29

page 103

A science student is riding on a flatcar of a train traveling
along a straight, horizontal track at a constant speed
of 10.0 m/s. The student throws a ball into the air along
a path that he judges to make an initial angle of 60.0°
with the horizontal and to be in line with the track.
The student’s professor, who is standing on the ground
nearby, observes the ball to rise vertically. How high
does she observe the ball rise?

<img src="..//pics/Fig_ch4_problem_29.png" width=300> 


```python
# Given
v_train_x = 10.0 * (u.m / u.s)
theta = vp.radians(60.0)

# For the professor to see it vertically, the ball's horizontal
# velocity relative to the train must be -10 m/s
v_bt_x = -v_train_x

# v_bt_x = v0 * cos(theta) ==> v0 = v_bt_x / cos(theta)
v0_mag = abs(v_bt_x / vp.cos(theta))

# Vertical component
v_iy = v0_mag * vp.sin(theta)

# Max Height H = vy^2 / (2g)
g = 9.8 * (u.m / u.s**2)
h_max = (v_iy**2) / (2 * g)

print(f"Initial throw speed relative to train: {v0_mag:.2f}")
print(f"Vertical velocity component: {v_iy:.2f}")
print(f"Maximun height observed: {h_max:.2f}")
```

#### Problem 30

page 103

A river has a steady speed of 0.500 m/s. A student swims
upstream a distance of 1.00 km and swims back to the
starting point. 

**(a)** 

If the student can swim at a speed
of 1.20 m/s in still water, how long does the trip take?

```python
v_river = 0.500 * (u.m / u.s)
distance = 1_000 * u.m
v_student = 1.20 * (u.m / u.s)

# Velocity up the river
v_up = v_student - v_river
# Time to go up the river 1 km
# v = D / t ==> t = D / v
t_up = distance / v_up

# Velocity down the river
v_down = v_student + v_river
# Time to go down the river 1 km
t_down = distance / v_down

# Total time
t_total = t_up + t_down

print(f"Time to go up: {t_up:.1f} equivalent to: {t_up.to(u.min):.1f}")
print(f"Time to go down: {t_down:.1f} equivalent to: {t_down.to(u.min):.1f}")
print(f"Total time: {t_total:.1f} equivalent to: {t_total.to(u.min):.1f}")
```

**(b)** 

How much time is required in still water for the
same length swim? 

```python
# Time in still water
t_still = 2 * distance / v_student
print(
    f"Total time in still water: {t_still:.1f} equivalent to: {t_still.to(u.min):.1f}"
)
```

**(c)** 

Intuitively, why does the swim
take longer when there is a current?

**Answer:**

The student moves slower upstream, they spend a longer time fighting the current
that they do being helped by it going downstream.

The "penalty" of the current last longer than the "boost", resulting a net loss 
of time compared to still water.

You are participating in a summer internship with the
Coast Guard. You have been assigned the duty of determining
the direction in which a Coast Guard speedboat
should travel to intercept unidentified vessels. One day,
the radar operator detects an unidentified vessel at a distance
of 20.0 km from the radar installation in the direction
15.0° east of north. The vessel is traveling at 26.0 km/h
on a course at 40.0° east of north. The Coast Guard wishes
to send a speedboat, which travels at 50.0 km/h, to travel
in a straight line from the radar installation to intercept
and investigate the vessel, and asks you for the heading for
the speedboat to take. Express the direction as a compass
bearing with respect to due north.

```python
v_vessel_speed = 26.0 * (u.km / u.h)
v_speedboat = 50.0 * (u.km / u.h)

# Angle Shift: Because the x-axis is now pointed at 15 deg, the vessel's travel
# direction 40 deg. is now an angle of 40 - 15 = 25 deg. relative to the new x-axis
theta = vp.radians(40 - 15)

# Choose the x-axis along the 20-km distance. The y components of the
# displacements of the ship and the speed boat must agree.
# Since the vessel starts at y=0, its vertical displacement after time t is:
# y_v = (v_vessel sin(theta) t
# The speed boat also starts at y=0
# y_s = (v_speed sin(alpha) t
# setting y_v = y_s allows the t to cancel out
# (v_vessel sin(theta) = (v_speed sin(alpha)
# ==> sin(alpha) = v_vessel sin(theta) / v_speed
alpha_rad = vp.asin(v_vessel_speed * vp.sin(theta) / v_speedboat)
alpha_deg = vp.degrees(alpha_rad)
# For the bearing we must compensate the angle shift of 15 deg
bearing = 15 + alpha_deg
print(f"The speedboat should head bearing={bearing:.1f} deg.")
```

of 9.50 m/s on a limitless, horizontal stretch of road.

**(a)** 

In the frame of reference
of the truck, at what angle to the vertical does the boy

**Answer:**

As the boy catches the can at the same location on the truck bed, it means the 
can had zero horizontal displacement relative to the truck. The can must have
been thrown perfectly straight up.

* **Angle to the vertical:** 0° (straight up.)

**(b)** 

What is the initial speed of the can
relative to the truck? 


```python
v_truck = 9.50 * (u.m / u.s)
delta_x = 16.0 * u.m
g = 9.8 * (u.m / u.s**2)

# The truck travels 16.0 at a constant velocity of 9.50 m/s^2 during the time
# the can is in the air. t = delta_x / v_truck

t = delta_x / v_truck

# the can's vertical motion is governed by gravity. It reaches the peak at t/2
# and returns. From the kinematic equation v_f = v_i + a t; for the trip back
# down (where v_i = 0 at the peak) ==> vi = g /

# Free fall equation for the can yf = yi + v_yi t - (1/2)g t^2
# yf = yi = 0 # starts and finishes and the hand of the boy
# v_yi = (1/2)g t^2
v_yi = (1 / 2) * g * t
print(f"The initial speed of the can relative to the truck v_yi={v_yi:.2f}")
```
