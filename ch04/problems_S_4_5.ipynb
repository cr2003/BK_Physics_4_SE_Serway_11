**(a)** 

Can a particle moving with instantaneous speed
3.00 m/s on a path with radius of curvature 2.00 m have
an acceleration of magnitude 6.00 m/s2? 


```python
v = 3.00 * (u.m / u.s)
r = 2.00 * u.m
a_t = 6.00 * (u.m / u.s**2)

a_c = v**2 / r
print(f"The particle's centripetal acceleration is a_c = {a_c}")

# Solving for tangential acceleration
# a_t^2 = a_c^2 + a_tg^2 ==> a_tg = (a_t^2 - a_c)^2
a_tg = (a_t**2 - a_c**2) ** (1 / 2)
print(
    f"To have a total acceleration a_t = {a_t}, could have a tangential acceleration with magnitude a_tg = {a_tg:.2f}"
)
print("Either speeding up or slowing down.")
```

**(b)** 

determine the magnitude of its radial acceleration

```python
a_x = 22.5 * (u.m / u.s**2)
a_y = 20.2 * (u.m / u.s**2)

# The components of the acceleration along the rope is the centripetal acceleration
# From the picture, note how the angles apply to the components into the rope
a_c = a_x * (cos(radians(90 - 36.9))) + a_y * (cos(radians(36.9)))
print(f"The centripetal acceleration a_c = {a_c:.1f}")
```

(c) determine the speed and velocity of the ball.

```python
r = 1.50 * u.m
# Speed
# a_c = v^2 / r ==> v = (a_c * r)^(1/2)
v = (a_c * r) ** (1 / 2)
print(f"The speed is v={v:.2f}")
print("The velocity is tangent to the circle")
```
