The astronaut orbiting the Earth in Figure P4.19 is preparing to dock with
the Westar 6 satellite. The satellite is in a circular orbit 600 km above the
Earth’s surface, where the free-fall acceleration is 8.21 m/s2. Take the radius 
of the Earth as 6 400 km. Determine the speed of the satellite and the time 
interval required to complete one orbit around the Earth, which is the period 
of the satellite.

```python
# Input values using Astropy Units
h = 600 * u.km
h = h.to(u.m)
# Free-fall aceleration at altitude h
g_at_h = 8.21 * (u.m / u.s**2)

# Earth Radius as given n the problem
R_E = 6_400 * u.km
R_E = R_E.to(u.m)

# Total Radius (r)
r = R_E + h

# Orbital Speed (v)
# a_c = v**2 / r ==> v = (a_c / r)**(1/2)
v = (g_at_h * r) ** (1 / 2)
print(f"Orbital speed v={v:.0f}")

# Orbital Period (T)
T = (2 * pi * r) / v
print(f"Orbital Period T={T:.0f} ~ {T.to(u.min):.0f}")
```

#### Problem 20

An athlete swings a ball, connected to the end of a chain,
in a horizontal circle. The athlete is able to rotate the ball
at the rate of 8.00 rev/s when the length of the chain is
0.600 m. When he increases the length to 0.900 m, he is
able to rotate the ball only 6.00 rev/s. 

**(a)**

 Which rate of rotation gives the greater speed for the ball? 

```python
omega_1 = 8.00 * (u.cycle / u.s)
omega_1 = omega_1.to(u.rad / u.s)
r_1 = 0.600 * u.m
omega_2 = 6.00 * (u.cycle / u.s)
omega_2 = omega_2.to(u.rad / u.s)
r_2 = 0.900 * u.m


v_1 = omega_1 * r_1
# Clean the rad
v_1 = (omega_1 * r_1) / u.rad
v_2 = omega_2 * r_2
# Clean the rad
v_2 = (omega_2 * r_2) / u.rad

print(
    f"For the length of chain r_1 = {r_1} with rotation rate of omega_1 = {omega_1:.2f} the speed of the ball is v_1 = {v_1:.2f}"
)
print(
    f"For the length of chain r_2 = {r_2} with rotation rate of omega_2 = {omega_2:.2f} the speed of the ball is v_2 = {v_2:.2f}"
)

print(f"So the rate of rotation 6.00 rev/s gives the greater speed of the ball")
```

**(b)**

What is the centripetal acceleration of the ball at 8.00 rev/s?


```python
# a_c = v^2 / r
a_c_8 = v_2**2 / r_2
print(f"Centripetal acceleration a_c_8 = {a_c_8:.0f}")
```

#### Problem 22

page 102

A tire 0.500 m in radius rotates at a constant rate of 200 rev/min. Find the
speed and acceleration of a small stone lodged in the tread of the tire (on its
outer edge).

```python
r = 0.500 * u.m
omega = 200 * (u.cycle / u.min)
omega = omega.to(u.rad / u.s)

# Speed (clean the radians)
v = (omega * r) / u.rad
print(f"Speed v = {v:.1f}")

# Acceleration
a_c = v**2 / r
print(f"Acceleration a_c = {a_c:.0f}")
```
