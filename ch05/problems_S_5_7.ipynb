Figure P5.15 shows a
worker poling a boat—a very efficient
mode of transportation—
across a shallow lake. He pushes
parallel to the length of the light
pole, exerting a force of magnitude
240 N on the bottom of the
lake. Assume the pole lies in the
vertical plane containing the
keel of the boat. At one moment,
the pole makes an angle of 35.0°
with the vertical and the water
exerts a horizontal drag force of
47.5 N on the boat, opposite to its
forward velocity of magnitude 0.857 m/s. The mass of
the boat including its cargo and the worker is 370 kg.

**(a)** The water exerts a buoyant force vertically upward
on the boat. Find the magnitude of this force. 

**(b)** Model
the forces as constant over a short interval of time to
find the velocity of the boat 0.450 s after the moment
described.



```python
theta_deg = 35 * u.deg
theta_rad = theta_deg.to(u.rad)
# Pole force on the boat magnitude (F_pb_mag)
F_pb_mag = 240 * u.N
# Water drag force
F_wd_mag = 47.5 * u.N
mass = 370 * u.kg
g = -9.8 * (u.m / u.s**2)

# Pole force on the boat (F_pb)
# As theta is meassured from vertical (sin and cos are inverted)
F_pb_vec = vp.vec(
    F_pb_mag.value * vp.sin(theta_rad.value),
    F_pb_mag.value * vp.cos(theta_rad.value),
    0,
)

# Gravitational Force F_g
F_g = mass * g
F_g_vec = vp.vec(0, F_g.value, 0)

# Calculate Buoyant force F_b in the y-axis
F_b = -(F_g_vec.y + F_pb_vec.y) * u.N
print("(a)")
print(f"The buoyant force is F_b = {F_b:,.0f}")
```

```python
print("(b)")
# Force in the x-axis, water drag is negative as opposites the avance in the x-axis
F_x = (F_pb_vec.x * u.N) - F_wd_mag
# F_x = m a_x ==> a_x = F_x / m
a_x = (F_x / mass).to(u.m / u.s**2)

# v_xf = v_xi * a_x * t
t = 0.450 * u.s
v_xi = 0.857 * (u.m / u.s)
v_xf = v_xi + a_x * t
print(f"Velocity of board at this moment v_xf={v_xf:.3f}")
```

An iron bolt of mass 65.0 g hangs from a string
35.7 cm long. The top end of the string is fixed. Without
touching it, a magnet attracts the bolt so that it remains
stationary, but it is displaced horizontally 28.0 cm to
the right from the previously vertical line of the string.
The magnet is located to the right of the bolt and on the
same vertical level as the bolt in the final configuration.

**(a)** Draw a free-body diagram of the bolt. 

**(b)** Find the tension in the string. 

**(c)** Find the magnetic force on the
bolt.

```python
# Checking the left-hand diagram, we find the angle theta
mass = (65.0 * u.g).to(u.kg)
g_mag = 9.80
g = g_mag * (u.m / u.s**2)
theta_rad = vp.acos(28 / 35.7) * u.rad
theta_deg = theta_rad.to(u.deg)
print(f"The angle theta = {theta_deg:.1f}")
# weight of the bolt
w = (mass * g).to(u.N)
print(f"The weight of the bolt w = {w:.3f}")
```

```python
print("(b)")
# To find the tension in the string, applying Newton's second law in x and y directions
# eq1: F_x = m * a_x ==> -T cos(theta) + F_mag = 0 # zero because equilibrium so 0 acceleration
# eq2: F_y = m * a_y ==> T sin(theta) - w = 0 # zero because equilibrium so 0 acceleration
# From equation 2 ==> T = w / sin(theta)
T = w / np.sin(theta_rad)
print(f"Tension T={T:.2f}")
```

```python
print("(c)")
# Now from eq1:
F_mag = T * np.cos(theta_rad)
print(f"Magnetic force F_mag={F_mag:.3f}")
```

A block slides down a frictionless plane having an inclination
of $\theta = 15.0°$. The block starts from rest at the
top, and the length of the incline
is 2.00 m. 

**(a)** Draw a free-body
diagram of the block. Find 

**(b)** the acceleration of the block and

**(c)** its speed when it reaches the
bottom of the incline.

(a)
<img src="..//pics/FigP5_17.png" width=400> 
NB: To solve the problem I take the positive x following the incline to the left

```python
print("(b)")
theta = 15.0 * u.deg
v_0 = 0 * (u.m / u.s)
x = 2.00 * u.m
g = 9.8 * (u.m / u.s**2)

# eq1: Fx = m g sin(theta)
# eq2: Fx = m a_x
# equating the two equations, mass cancels out in both equations
# a_x = g sin(theta)
a_x = g * np.sin(theta)  # np takes care of the degrees and astropy
print(f"Block acceleration a_x = {a_x:.2f}")
```

```python
print("(c)")
# v_f^2 = v_0^2 + 2 a x ==> v_f = sqrt(v_0^2 + 2 a x )
v_f = np.sqrt(v_0**2 + (2 * a_x * x))
print(f"Speed at end of incline v_f={v_f:.2f}")
```

The distance between two telephone poles is 50.0 m.
When a 1.00-kg bird lands on the telephone wire
midway between the poles, the wire sags 0.200 m.

**(a)** Draw a free-body diagram of the bird. 

**(b)** How much
tension does the bird produce in the wire? Ignore the
weight of the wire.

<img src="..//pics/FigP5_19.png" width=400> 


```python
x = (50 / 2) * u.m
y = (0.200) * u.m
mass = 1 * u.kg
g = 9.8 * (u.m / u.s**2)


alpha = np.atan2(y, x)
weight = mass * g

# Calculate the tension in the wire, T_x cancel out
# In the y-axis F_y(weight) = 2 T sin(alpha) # (Tension on the left and right hand of the wire)
#  weight = 2 T sin(alpha) ==> T = weight / 2 sin(alpha)
T = weight / (2 * np.sin(alpha))
print(f"The tension T = {T.to(u.N):.0f}")
```

An object of mass m1 = 5.00 kg
placed on a frictionless, horizontal
table is connected to a string that
passes over a pulley and then is fastened
to a hanging object of mass
m2 = 9.00 kg as shown in
Figure P5.22. 

**(a)** Draw free-body
diagrams of both objects. Find

**(b)** the magnitude of the acceleration
of the objects and 

**(c)** the tension in the string.




(a) Diagram

<img src="..//pics/FigP5_22_2.png" width=300> 



```python
# Define symbolic variables (positive real numbers)
m1_s, m2_s, g_s, a_s, T_s = sp.symbols("m_1 m_2 g a T", positive=True)

# Equations based on Newton's 2nd Law
# m1: T = m1 * a
# m2: m2 * g - T = m2 * a
eq1 = sp.Eq(T_s, m1_s * a_s)
eq2 = sp.Eq(m2_s * g_s - T_s, m2_s * a_s)

# Solve system of linear equations for 'a' and 'T'
sol = sp.solve([eq1, eq2], (a_s, T_s))

# Store symbolic expressions for use in Script 2
expr_a = sol[a_s]
expr_T = sol[T_s]

print("--- SYMBOLIC FORMULAS ---")
print("Acceleration formula a =", expr_a)
print("Tension formula T      =", expr_T)
```

```python
# Convert SymPy expressions into callable functions
func_a = sp.lambdify((m1_s, m2_s, g_s), expr_a, modules="numpy")
func_T = sp.lambdify((m1_s, m2_s, g_s), expr_T, modules="numpy")

# Given AstroPy values
m1_val = 5.00 * u.kg
m2_val = 9.00 * u.kg
g_val = 9.80 * (u.m / u.s**2)

# Substitute values into lambdified functions
a_result = func_a(m1_val, m2_val, g_val)
T_result = func_T(m1_val, m2_val, g_val)

print("--- EVALUATED RESULTS (ASTROPY) ---")
print(f"(b) Acceleration magnitude: {a_result:.1f}")
print(f"(c) String tension:        {T_result.to(u.N):.1f}")
```
