#### Problem 9. Review.

page 135

The gravitational force exerted on a baseball
is 2.21 N down. A pitcher throws the ball horizontally
with velocity 18.0 m/s by uniformly accelerating it along
a straight horizontal line for a time interval of 170 ms.
The ball starts from rest. 

**(a)** Through what distance
does it move before its release? 

**(b)** What are the magnitude
and direction of the force the pitcher exerts on
the ball?



```python
w = 2.21 * u.N
g = 9.80 * u.m / u.s**2
v_f = 18.0 * (u.m / u.s)
v_i = 0 * (u.m / u.s)
delta_t = (170 * u.ms).to(u.s)

# w = m g ==> m = w / g
m = (w / g).to(u.kg)

print("(a)")
# delta_x = x_f - x_i
# x_f = x_i + (1/2)*(v_i + v_f)*delta_t ==> delta_x = (1/2)*(v_i + v_f) * delta_t
delta_x = (1 / 2) * (v_i + v_f) * delta_t
print(f"Delta x: {delta_x}")
```

```python
print("(b)")
a_x = (v_f - v_i) / delta_t
# F_1: Force of pitcher on the ball
# F_2: Force of Earth on the ball (weight)
# F_net = F_1 + F_2 = m a
# Using vPython
F_1 = vp.vec(m.value * a_x.value, w.value, 0)
F_1_mag = F_1.mag * u.N
print(f"F_1_mag = {F_1_mag:.1f}")
theta_rad = vp.atan2(F_1.y, F_1.x) * u.rad
theta_deg = theta_rad.to(u.deg)
theta_deg
print(f"theta = {theta_deg:.2f}")
print(
    f"The pitcher exerts a force of {F_1_mag:.1f} at {theta_deg:.2f} above the horizontal."
)
```

#### Problem 12

page 135

If an astronaut weighs 900 N on the Earth, what would
they weigh in their spacecraft floating near the cloud tops
of Jupiter, where the free-fall acceleration is 25.9 m/s2?

```python
w_earth = 900 * u.N
g_earth = 9.80 * (u.m / u.s**2)
g_jupyter = 25.9 * (u.m / u.s**2)

m = w_earth / g_earth
m.to(u.kg)
w_jupiter = (m * g_jupyter).to(u.N)
print(f"The astronaut weigh in Jupiter: {w_jupiter:,.0f}")
```

```python
print("(a)")
# You and Earth exert equal forces on each other
# m_y * g = M_e * a_e ==> a_E = (m_y * g) / M_e
m_y = 70 * u.kg
g = const.g0
M_e = const.M_earth

a_e = (m_y * g) / M_e
print(f"Earth acceleration order of magnitude {oom(a_e)}")
```

```python
print("(b)")
# You and the planet move for equal time interval delta_t
# delta_x = (1/2)*a*(delta_t)^2 ==> delta_t = sqrt((2*delta_x)/a)
delta_x_y = 0.5 * u.m  # high of the seat
delta_x_earth = (a_e / g) * delta_x_y

print(f"Earth motion order of magnitude {oom(delta_x_earth)}")
```
