Why is the following situation impossible? A 3.80-kg
book is placed next to you on the horizontal seat of your
car. The coefficient of static friction between the book
and the seat is 0.650, and the coefficient of kinetic friction
is 0.550. You are traveling forward at 72.0 km/h
and brake to a stop with constant acceleration over a
distance of 30.0 m. The book remains on the seat rather
than sliding forward onto the floor.



```python
m_b = 3.80 * u.kg
mu_s = 0.650
mu_k = 0.550
v_i = 72.0 * (u.km / u.h)
v_i = v_i.to(u.m / u.s)
v_f = 0 * (u.m / u.s)
delta_x = 30.0 * u.m
g = 9.8 * (u.m / u.s**2)

# v_f^2 = v_i^2 + 2 a delta_x ==> a = (-vi^2 / 2 delta_x)
a_x = (-(v_i**2)) / (2 * delta_x)

# normal force applied to the book
n = m_b * g

# Maximum acceleration that static friction can stop the book
f_s = mu_s * n
# f = m a ==> a = f/m ==> a_max = f_s / m_b
a_max = f_s / m_b
a_max
print(f"The maximum acceleration static friction can stop the book a_max = {a_max:.2f}")
print(f"The car acceleration due to brake a_x = {a_x:.2f}")
print(
    "Maximun static friction cannot provide enough acceleration to keep the book stationary in the seat"
)
```

<img src="..//pics/FigP5_22.png" width=300> 

A 9.00-kg hanging object is connected by a light, inextensible
cord over a light, frictionless pulley to a 5.00-kg
block that is sliding on a flat table (Fig. P5.22). Taking
the coefficient of kinetic friction as 0.200, find the tension
in the string.




```python
# Define symbolic variables (positive real numbers)
m1_s, m2_s, g_s, muk_s, a_s, T_s = sp.symbols("m_1 m_2 g mu_k a T", positive=True)

# Equations based on Newton's 2nd Law with friction
# m1: T - mu_k * m1 * g = m1 * a
# m2: m2 * g - T = m2 * a
eq1 = sp.Eq(T_s - muk_s * m1_s * g_s, m1_s * a_s)
eq2 = sp.Eq(m2_s * g_s - T_s, m2_s * a_s)

# Solve system of linear equations for 'a' and 'T'
sol = sp.solve([eq1, eq2], (a_s, T_s))

# Store symbolic expressions
expr_a = sol[a_s]
expr_T = sol[T_s]

print("--- SYMBOLIC FORMULAS ---")
print("Acceleration formula a =", sp.simplify(expr_a))
print("Tension formula T      =", sp.simplify(expr_T))
```

```python
# Convert SymPy expressions into callable functions
func_a = sp.lambdify((m1_s, m2_s, g_s, muk_s), expr_a, modules="numpy")
func_T = sp.lambdify((m1_s, m2_s, g_s, muk_s), expr_T, modules="numpy")

# Given values
m1_val = 5.00 * u.kg
m2_val = 9.00 * u.kg
g_val = 9.80 * (u.m / u.s**2)
muk_val = 0.200

# Evaluate via lambdified functions
a_result = func_a(m1_val, m2_val, g_val, muk_val)
T_result = func_T(m1_val, m2_val, g_val, muk_val)

print("--- EVALUATED RESULTS (ASTROPY) ---")
print(f"Acceleration magnitude: {a_result:.1f}")
print(f"Tension in string:      {T_result.to(u.N):.1f}")
```

#### Problem 33

You have been called as an expert witness for a trial
in which a driver has been charged with speeding but
is claiming innocence. He claims to have slammed on
his brakes to avoid rear-ending another car, but tapped
the back of the other car just as he came to rest. You
have been hired by the prosecution to prove that the
driver was indeed speeding. You have received data as
follows from the police: Skid marks left by the driver
are 56.0 m long and the roadway is level. Tires matching
those on the car of the driver have been dragged
over the same roadway to determine that the coefficient
of kinetic friction between the tires and the roadway
is 0.82 at all points along the skid mark. The speed
limit on the road is 35 mi/h. Construct an argument
to be used in court to show that the driver was indeed
speeding.


```python
# 1. Define symbolic variables (leave ax_s without 'positive=True' since deceleration is negative)
g_s, muk_s, d_s, vi_s = sp.symbols("g mu_k d v_i", positive=True)
ax_s = sp.Symbol("a_x", real=True)

# 2. Newton's 2nd Law:
# When the tires lock, kinetic friction is the only horizontal force acting on the car.
# Setting $+x$ in the direction of motion:
# -f_k = m a_x
# f_k = muk n = muk m g
# m a_x = - muk m g  # m cancel out
# a_x = -mu_k * g
ax_expr = -muk_s * g_s

# 3. Kinematics Equation: v_f^2 = v_i^2 + 2 * a_x * d (with v_f = 0)
# 0 = v_i^2 + 2 * (-mu_k * g) * d
kinematic_eq = sp.Eq(0, vi_s**2 + 2 * ax_expr * d_s)

# 4. Solve for v_i
sol_vi = sp.solve(kinematic_eq, vi_s)[0]

print("--- SYMBOLIC FORMULAS ---")
print("Braking acceleration a_x =", ax_expr)
print("Initial speed v_i        =", sol_vi)
```

```python
# Convert SymPy expression to a callable function
func_vi = sp.lambdify((muk_s, g_s, d_s), sol_vi, modules="numpy")

# Given values
muk_val = 0.82
g_val = 9.80 * (u.m / u.s**2)
d_val = 56.0 * u.m
speed_limit = 35.0 * (u.imperial.mile / u.h)

# Compute initial speed
vi_result = func_vi(muk_val, g_val, d_val)
vi_mph = vi_result.to(u.imperial.mile / u.h)

print("--- EVALUATED RESULTS (ASTROPY) ---")
print(f"Braking Deceleration:           {(-muk_val * g_val):.2f}")
print(f"Calculated Minimum Speed (v_i): {vi_result:.2f} ({vi_mph:.2f})")
print(f"Posted Speed Limit:             {speed_limit:.2f}")
print(f"Over Speed Limit By:            {(vi_mph - speed_limit):.2f}")
```

A block of mass 3.00 kg
is pushed up against
a wall by a force $\vec{P}$
that makes an angle
of $\theta = 50.0°$ with the
horizontal as shown in
Figure P5.34. The coefficient
of static friction
between the block and the wall is 0.250. 

**(a)** Determine
the possible values for the magnitude of P S that allow the
block to remain stationary. 

**(b)** Describe what happens if
|$\vec{P}$| has a larger value and what happens if it is smaller.

**(c)** Repeat parts (a) and (b), assuming the force makes
an angle of $\theta = 13.0°$ with the horizontal.




```python
import sympy as sp

# Define symbolic variables
m_s, g_s, L_s = sp.symbols("m g L", positive=True)
v1_s, v2_s, P_s = sp.symbols("v_1 v_2 P", positive=True)

# Accelerations in each phase
a1_s = P_s / m_s
a2_s = (P_s - m_s * g_s) / m_s

# Kinematic equation over distance L (two halves L/2)
# v2^2 - v1^2 = 2 * a1 * (L/2) + 2 * a2 * (L/2)
eq_kinematics = sp.Eq(v2_s**2 - v1_s**2, a1_s * L_s + a2_s * L_s)

# Solve for thrust force P
sol_P = sp.solve(eq_kinematics, P_s)[0]

print("--- SYMBOLIC FORMULA ---")
print("Constant thrust force P =", sp.simplify(sol_P))
```

```python
# Convert SymPy expressions to callable functions
func_Pa = sp.lambdify((m_s, g_s, theta_s, mus_s), sol_Pa, modules="numpy")
func_Pb = sp.lambdify((m_s, g_s, theta_s, muk_s), sol_Pb, modules="numpy")
func_Pc = sp.lambdify((m_s, g_s, theta_s, mus_s), sol_Pc, modules="numpy")

# Given values from problem statement (with theta = 50 degrees)
m_val = 3.00 * u.kg
g_val = 9.80 * (u.m / u.s**2)
theta_val = np.radians(50.0)  # Convert 50 degrees to radians
mus_val = 0.250
muk_val = 0.150

# Evaluate forces
P_a_res = func_Pa(m_val, g_val, theta_val, mus_val).to(u.N)
P_b_res = func_Pb(m_val, g_val, theta_val, muk_val).to(u.N)
P_c_res = func_Pc(m_val, g_val, theta_val, mus_val).to(u.N)

print("--- EVALUATED RESULTS (ASTROPY) ---")
print(f"(a) P to initiate motion UP:     {P_a_res:.1f}")
print(f"(b) P for constant speed UP:      {P_b_res:.1f}")
print(f"(c) P to prevent sliding DOWN:   {P_c_res:.1f}")
```

**(c)** Repeat parts (a) and (b), assuming the force makes

A Chinook salmon can swim underwater at
3.58 m/s, and it can also jump vertically upward, leaving
the water with a speed of 6.26 m/s. A record salmon
has length 1.50 m and mass 61.0 kg. Consider the fish
swimming straight upward in the water below the surface
of a lake. The gravitational force exerted on it is
very nearly canceled out by a buoyant force exerted by
the water as we will study in Chapter 14. The fish experiences
an upward force P exerted by the water on its
threshing tail fin and a downward fluid friction force
that we model as acting on its front end. Assume the
fluid friction force disappears as soon as the fish’s head
breaks the water surface and assume the force on its tail
is constant. Model the gravitational force as suddenly
switching full on when half the length of the fish is out
of the water. Find the value of P.



```python
# Define symbolic variables
m_s, g_s, L_s = sp.symbols("m g L", positive=True)
v1_s, v2_s, P_s = sp.symbols("v_1 v_2 P", positive=True)

# Emerging Stage 1 (Head breaks surface to half-length (L/2))
# Emerging Stage 2 (Half-length to fully out of water (L/2))
# Accelerations in each phase
a1_s = P_s / m_s
a2_s = (P_s - m_s * g_s) / m_s

# Kinematic equation over distance L (two halves L/2)
# v2^2 - v1^2 = 2 * a1 * (L/2) + 2 * a2 * (L/2)
eq_kinematics = sp.Eq(v2_s**2 - v1_s**2, a1_s * L_s + a2_s * L_s)

# Solve for thrust force P
sol_P = sp.solve(eq_kinematics, P_s)[0]

print("--- SYMBOLIC FORMULA ---")
print("Constant thrust force P =", sp.simplify(sol_P))
```

```python
# Convert SymPy expression into a callable function
func_P = sp.lambdify((m_s, g_s, L_s, v1_s, v2_s), sol_P, modules="numpy")

# Given values from problem statement
m_val = 61.0 * u.kg
g_val = 9.80 * (u.m / u.s**2)
L_val = 1.50 * u.m
v1_val = 3.58 * (u.m / u.s)
v2_val = 6.26 * (u.m / u.s)

# Calculate force P
P_result = func_P(m_val, g_val, L_val, v1_val, v2_val)
P_result = P_result.to(u.N)

print("--- EVALUATED RESULTS (ASTROPY) ---")
print(f"Upward Tail Force P: {P_result:.0f}")
```

A 5.00-kg block is placed on top of a 10.0-kg
block (Fig. P5.36). A horizontal force of
45.0 N is applied to the 10-kg block, and
the 5.00-kg block is tied to the wall. The
coefficient of kinetic friction between all moving surfaces
is 0.200. 

**(a)** Draw a free-body diagram for each
block and identify the action–reaction forces between
the blocks. 

**(b)** Determine the tension in the string and
the magnitude of the acceleration of the 10.0-kg block.


**(a)** The free-body diagrams are shown in the figure below.

<img src="..//pics/FigP5_36_2.png" width=600> 


```python
print("(b)")

# 1. Define symbolic variables
m1, m2, F, mu_k, g = sp.symbols("m1 m2 F mu_k g", positive=True)
T, a = sp.symbols("T a", real=True)

# 2. Normal forces
N1 = m1 * g
N2 = (m1 + m2) * g

# 3. Kinetic friction forces
f_k1 = mu_k * N1
f_k2 = mu_k * N2

# 4. Equations of motion
# For top block (m1): T - f_k1 = 0
eq1 = sp.Eq(T, f_k1)

# For bottom block (m2): F - f_k1 - f_k2 = m2 * a
eq2 = sp.Eq(F - f_k1 - f_k2, m2 * a)

# 5. Solve symbolically
sol_T = sp.solve(eq1, T)[0]
sol_a = sp.solve(eq2, a)[0]

print("Symbolic Expression for Tension (T):")
sp.pprint(sol_T)

print("\nSymbolic Expression for Acceleration (a):")
sp.pprint(sol_a)
```

```python
# 1. Convert SymPy expressions to callable functions using lambdify
func_T = sp.lambdify((m1, mu_k, g), sol_T, "numpy")
func_a = sp.lambdify((m1, m2, F, mu_k, g), sol_a, "numpy")

# 2. Define real value variables with Astropy units
m1_val = 5.00 * u.kg
m2_val = 10.0 * u.kg
F_val = 45.0 * u.N
mu_k_val = 0.200
g_val = const.g0  # Standard acceleration due to gravity (~9.80665 m/s^2)

# 3. Evaluate functions (extract magnitude using .value and append units)
T_num = func_T(m1_val.value, mu_k_val, g_val.value) * u.N
a_num = func_a(m1_val.value, m2_val.value, F_val.value, mu_k_val, g_val.value) * (
    u.m / u.s**2
)

# 4. Display results
print(f"Tension in the string, T = {T_num:.2f}")
print(f"Acceleration of 10.0-kg block, a = {a_num:.3f} ")
```
